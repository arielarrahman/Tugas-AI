# DEMONSTRASI PENGOLAHAN MENGGUNAKAN DATA SIMULASI

**Catatan penting untuk pelaporan:** notebook asli berisi rancangan prompt untuk dua metode, tetapi tidak menyertakan data pengukuran lapangan. Agar alur pengolahan dan pembuatan gambar dapat dicoba, bagian berikut membuat **data simulasi (data sintetis) yang dibuat oleh kode**, bukan hasil observasi lapangan. Nilai-nilai ini boleh digunakan untuk demonstrasi/pengujian kode, tetapi jangan dilaporkan sebagai data pengukuran nyata. Jika tugas mensyaratkan data lapangan, data simulasi perlu diganti dengan data hasil pengukuran atau dataset yang disetujui dosen.

Data dibuat secara reprodusibel dengan random seed tetap, sehingga hasilnya tetap sama setiap kali notebook dijalankan. Notebook akan menampilkan tabel data, analisis awal, grafik, dan mengekspor berkas CSV serta PNG.

```python
# 1. Persiapan pustaka
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display

# Folder keluaran: folder kerja tempat notebook dijalankan
OUTPUT_DIR = Path.cwd()
print("Folder keluaran:", OUTPUT_DIR.resolve())
print("Seluruh data di bawah ini adalah DATA SIMULASI, bukan pengukuran lapangan.")
```

## 1. Metode Seismik Refraksi

Untuk latihan ini dibuat model ideal dua lapisan horizontal. Waktu tiba pertama disimulasikan sebagai waktu minimum antara gelombang langsung pada lapisan pertama dan gelombang refraksi kepala pada lapisan kedua. Gangguan acak kecil ditambahkan agar data tidak sepenuhnya ideal.

**Parameter model untuk membuat simulasi (bukan hasil estimasi):** kecepatan lapisan 1 = 500 m/s, kecepatan lapisan 2 = 1.800 m/s, dan kedalaman antarmuka = 4 m. Nilai ini merupakan asumsi pembuat simulasi semata.

```python
# 2. Membuat data simulasi seismik refraksi
rng_seismic = np.random.default_rng(2026)

# Posisi geofon (m), dengan sumber/shot point pada x = 0 m
x_m = np.arange(2.0, 32.0, 2.0)
v1_model_m_s = 500.0
v2_model_m_s = 1800.0
h_model_m = 4.0

critical_angle_model = np.arcsin(v1_model_m_s / v2_model_m_s)
intercept_time_model_s = 2.0 * h_model_m * np.cos(critical_angle_model) / v1_model_m_s

time_direct_s = x_m / v1_model_m_s
time_head_s = intercept_time_model_s + x_m / v2_model_m_s
time_first_arrival_s = np.minimum(time_direct_s, time_head_s)
noise_s = rng_seismic.normal(loc=0.0, scale=0.00022, size=len(x_m))
time_observed_ms = (time_first_arrival_s + noise_s) * 1000.0

# Waktu tiba tidak boleh negatif (pengamanan sederhana)
time_observed_ms = np.maximum(time_observed_ms, 0.01)

data_seismik = pd.DataFrame({
    "Geofon": [f"G{i:02d}" for i in range(1, len(x_m) + 1)],
    "Posisi_Geofon_m": x_m,
    "Jarak_dari_Sumber_m": x_m,
    "Waktu_Tiba_Pertama_ms": np.round(time_observed_ms, 3),
})

csv_seismik = OUTPUT_DIR / "data_simulasi_seismik_refraksi.csv"
data_seismik.to_csv(csv_seismik, index=False)
print("Data simulasi seismik refraksi disimpan:", csv_seismik.name)
display(data_seismik)
print("Jumlah titik:", len(data_seismik))
```

### Analisis kurva waktu tiba

Kemiringan kurva waktu–jarak berhubungan dengan kebalikan kecepatan (slope = 1/v). Untuk demonstrasi model dua lapisan, titik dengan jarak sampai 10 m digunakan sebagai cabang gelombang langsung, sedangkan titik mulai 12 m digunakan sebagai cabang refraksi pada model ini. Pembagian cabang tersebut mengikuti skenario simulasi dan harus ditinjau ulang jika menggunakan data nyata.

```python
# 3. Estimasi kecepatan dari kemiringan dua cabang kurva
x = data_seismik["Jarak_dari_Sumber_m"].to_numpy(dtype=float)
t_s = data_seismik["Waktu_Tiba_Pertama_ms"].to_numpy(dtype=float) / 1000.0

mask_langsung = x <= 10.0
mask_refraksi = x >= 12.0
if mask_langsung.sum() < 2 or mask_refraksi.sum() < 2:
    raise ValueError("Titik untuk regresi cabang gelombang tidak cukup.")

slope1, intercept1 = np.polyfit(x[mask_langsung], t_s[mask_langsung], 1)
slope2, intercept2 = np.polyfit(x[mask_refraksi], t_s[mask_refraksi], 1)

v1_est_m_s = 1.0 / slope1
v2_est_m_s = 1.0 / slope2
intercept_est_s = intercept2

# Estimasi kedalaman menggunakan asumsi dua lapisan horizontal, V2 > V1
if v2_est_m_s > v1_est_m_s > 0:
    angle_est = np.arcsin(np.clip(v1_est_m_s / v2_est_m_s, 0, 1))
    h_est_m = intercept_est_s * v1_est_m_s / (2.0 * np.cos(angle_est))
else:
    angle_est = np.nan
    h_est_m = np.nan

hasil_seismik = pd.DataFrame({
    "Parameter": ["Kecepatan lapisan 1 (estimasi)", "Kecepatan lapisan 2 (estimasi)", "Waktu intersep cabang refraksi", "Kedalaman antarmuka (estimasi)"],
    "Nilai": [v1_est_m_s, v2_est_m_s, intercept_est_s * 1000.0, h_est_m],
    "Satuan": ["m/s", "m/s", "ms", "m"]
})
print("HASIL ESTIMASI DARI DATA SIMULASI")
display(hasil_seismik.round(3))

# Plot kurva waktu tiba dan garis regresi yang dipakai
fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(x, t_s * 1000.0, color="black", label="Waktu tiba simulasi", zorder=3)
xx1 = np.linspace(x[mask_langsung].min(), x[mask_langsung].max(), 100)
xx2 = np.linspace(x[mask_refraksi].min(), x[mask_refraksi].max(), 100)
ax.plot(xx1, (slope1 * xx1 + intercept1) * 1000.0, linewidth=2.2, label=f"Regresi gelombang langsung (V ≈ {v1_est_m_s:.0f} m/s)")
ax.plot(xx2, (slope2 * xx2 + intercept2) * 1000.0, linewidth=2.2, label=f"Regresi gelombang refraksi (V ≈ {v2_est_m_s:.0f} m/s)")
ax.set_title("Kurva Travel Time — Seismik Refraksi (Data Simulasi)")
ax.set_xlabel("Jarak dari sumber (m)")
ax.set_ylabel("Waktu tiba pertama (ms)")
ax.grid(True, alpha=0.3)
ax.legend()
fig.tight_layout()
png_seismik = OUTPUT_DIR / "kurva_travel_time_seismik_refraksi_simulasi.png"
fig.savefig(png_seismik, dpi=300, bbox_inches="tight")
plt.show()
print("Grafik disimpan:", png_seismik.name)
```

```python
# 4. Ringkasan interpretasi awal seismik refraksi
print("RINGKASAN INTERPRETASI — DATA SIMULASI")
print(f"Cabang awal menghasilkan estimasi kecepatan lapisan pertama sekitar {v1_est_m_s:.1f} m/s.")
print(f"Cabang jarak lebih jauh menghasilkan estimasi kecepatan lapisan kedua sekitar {v2_est_m_s:.1f} m/s.")
if np.isfinite(h_est_m):
    print(f"Dengan asumsi dua lapisan horizontal, kedalaman antarmuka yang dihitung sekitar {h_est_m:.2f} m.")
else:
    print("Kedalaman tidak dapat dihitung dengan rumus sederhana karena syarat V2 > V1 tidak terpenuhi.")
print("Interpretasi terbatas pada model dua lapisan ideal. Kecepatan saja tidak cukup untuk menetapkan jenis batuan, dan angka-angka ini tidak mewakili survei lapangan.")
```

## 2. Metode Geomagnetik

Karena data pengukuran magnetik dan koordinat lapangan tidak disertakan, bagian ini membuat grid koordinat lokal dan **anomali magnetik simulasi**. Pola dibuat dari satu zona anomali positif dan satu zona negatif berbentuk Gaussian, lalu ditambahkan noise kecil. Satuan anomali ditulis dalam nT untuk latihan visualisasi. Data ini tidak melalui koreksi variasi harian atau koreksi IGRF dan bukan data magnetometer lapangan.

```python
# 5. Membuat data simulasi anomali geomagnetik
rng_magnetik = np.random.default_rng(2027)
x_grid = np.arange(0.0, 71.0, 10.0)
y_grid = np.arange(0.0, 71.0, 10.0)
XX, YY = np.meshgrid(x_grid, y_grid)

anomali_positif = 85.0 * np.exp(-0.5 * (((XX - 25.0) / 16.0)**2 + ((YY - 30.0) / 17.0)**2))
anomali_negatif = -70.0 * np.exp(-0.5 * (((XX - 55.0) / 13.0)**2 + ((YY - 50.0) / 15.0)**2))
noise_magnetik = rng_magnetik.normal(0.0, 4.0, size=XX.shape)
anomali_nT = anomali_positif + anomali_negatif + noise_magnetik

# Tabel titik pengukuran simulasi dalam koordinat lokal (meter)
data_geomagnetik = pd.DataFrame({
    "Titik": [f"M{i:02d}" for i in range(1, XX.size + 1)],
    "Koordinat_X_lokal_m": XX.ravel(),
    "Koordinat_Y_lokal_m": YY.ravel(),
    "Anomali_Magnetik_Simulasi_nT": np.round(anomali_nT.ravel(), 2),
})
csv_magnetik = OUTPUT_DIR / "data_simulasi_geomagnetik.csv"
data_geomagnetik.to_csv(csv_magnetik, index=False)
print("Data simulasi geomagnetik disimpan:", csv_magnetik.name)
print("Contoh 10 baris pertama:")
display(data_geomagnetik.head(10))
print("Jumlah titik:", len(data_geomagnetik))
```

```python
# 6. Statistik ringkas dan peta anomali magnetik simulasi
idx_max = data_geomagnetik["Anomali_Magnetik_Simulasi_nT"].idxmax()
idx_min = data_geomagnetik["Anomali_Magnetik_Simulasi_nT"].idxmin()
row_max = data_geomagnetik.loc[idx_max]
row_min = data_geomagnetik.loc[idx_min]

stat_geomagnetik = pd.DataFrame({
    "Statistik": ["Minimum", "Maksimum", "Rata-rata", "Simpangan baku"],
    "Nilai (nT)": [data_geomagnetik["Anomali_Magnetik_Simulasi_nT"].min(),
                   data_geomagnetik["Anomali_Magnetik_Simulasi_nT"].max(),
                   data_geomagnetik["Anomali_Magnetik_Simulasi_nT"].mean(),
                   data_geomagnetik["Anomali_Magnetik_Simulasi_nT"].std()]
})
display(stat_geomagnetik.round(2))

fig, ax = plt.subplots(figsize=(9, 7))
contour = ax.contourf(XX, YY, anomali_nT, levels=18, cmap="RdBu_r")
ax.scatter(XX.ravel(), YY.ravel(), c=anomali_nT.ravel(), cmap="RdBu_r", edgecolor="black", linewidth=0.45, s=45)
ax.scatter(row_max["Koordinat_X_lokal_m"], row_max["Koordinat_Y_lokal_m"], marker="*", s=220, color="yellow", edgecolor="black", label="Anomali maksimum")
ax.scatter(row_min["Koordinat_X_lokal_m"], row_min["Koordinat_Y_lokal_m"], marker="X", s=120, color="lime", edgecolor="black", label="Anomali minimum")
ax.set_title("Peta Anomali Magnetik — Data Simulasi")
ax.set_xlabel("Koordinat X lokal (m)")
ax.set_ylabel("Koordinat Y lokal (m)")
ax.set_aspect("equal", adjustable="box")
ax.grid(True, alpha=0.15)
cb = fig.colorbar(contour, ax=ax)
cb.set_label("Anomali magnetik simulasi (nT)")
ax.legend(loc="upper right")
fig.tight_layout()
png_magnetik = OUTPUT_DIR / "peta_anomali_geomagnetik_simulasi.png"
fig.savefig(png_magnetik, dpi=300, bbox_inches="tight")
plt.show()
print("Peta disimpan:", png_magnetik.name)
print(f"Titik maksimum: ({row_max['Koordinat_X_lokal_m']:.0f} m, {row_max['Koordinat_Y_lokal_m']:.0f} m), nilai {row_max['Anomali_Magnetik_Simulasi_nT']:.2f} nT")
print(f"Titik minimum: ({row_min['Koordinat_X_lokal_m']:.0f} m, {row_min['Koordinat_Y_lokal_m']:.0f} m), nilai {row_min['Anomali_Magnetik_Simulasi_nT']:.2f} nT")
```

```python
# 7. Ringkasan interpretasi awal geomagnetik
print("RINGKASAN INTERPRETASI — DATA SIMULASI")
print(f"Rentang nilai anomali yang disimulasikan adalah {data_geomagnetik['Anomali_Magnetik_Simulasi_nT'].min():.2f} hingga {data_geomagnetik['Anomali_Magnetik_Simulasi_nT'].max():.2f} nT.")
print(f"Zona nilai relatif tinggi terletak di sekitar X={row_max['Koordinat_X_lokal_m']:.0f} m dan Y={row_max['Koordinat_Y_lokal_m']:.0f} m pada grid lokal.")
print(f"Zona nilai relatif rendah terletak di sekitar X={row_min['Koordinat_X_lokal_m']:.0f} m dan Y={row_min['Koordinat_Y_lokal_m']:.0f} m pada grid lokal.")
print("Perbedaan anomali dapat mengindikasikan variasi kemagnetan material bawah permukaan dalam konteks geofisika. Namun, bentuk peta ini dibuat secara sintetis; jenis litologi atau keberadaan sesar tidak dapat ditetapkan hanya dari peta ini tanpa data dan informasi geologi pendukung.")
```
