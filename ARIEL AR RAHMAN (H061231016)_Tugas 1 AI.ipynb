### **PROMPT 1 : ANALISIS DATA SEISMIK REFRAKSI**


#### **R — Role (Peran)**

Bertindak sebagai ahli Geofisika terapan dan dosen pembimbing yang memiliki keahlian dalam metode seismik refraksi. Memiliki kemampuan dalam pengolahan waktu tiba gelombang, analisis travel time, penentuan kecepatan gelombang P, interpretasi lapisan bawah permukaan, dan pembuatan model bawah permukaan. Lakukan analisis secara ilmiah dan sesuai tingkat pemahaman mahasiswa S1 Geofisika.

#### **T — Task (Tugas)**

1. Analisis data seismik refraksi untuk menentukan pola waktu tiba gelombang pada setiap geofon.
2. Identifikasi perubahan kemiringan kurva waktu tempuh terhadap jarak.
3. Estimasi kecepatan gelombang pada lapisan bawah permukaan berdasarkan data yang tersedia.
4. Interpretasikan kemungkinan susunan lapisan bawah permukaan berdasarkan hasil analisis.
5. Buat kesimpulan yang menjawab tujuan studi berdasarkan hasil pengolahan.

#### **C — Context (Konteks)**

Metode seismik refraksi memanfaatkan waktu tiba gelombang seismik yang direkam oleh geofon untuk mempelajari variasi kecepatan gelombang bawah permukaan. Perubahan kecepatan terhadap kedalaman dapat digunakan untuk memperkirakan batas antar lapisan. Interpretasi harus mempertimbangkan geometri survei, posisi sumber, jarak geofon, kualitas picking waktu tiba, dan informasi geologi yang tersedia.

#### **I — Input (Masukan)**

1. Data waktu tiba gelombang seismik.
2. Jarak atau posisi setiap geofon.
3. Posisi sumber atau shot point.
4. Grafik travel time jika tersedia.
5. Informasi topografi dan kondisi geologi lokasi penelitian jika tersedia.

#### **K — Constraints (Batasan)**

- Gunakan hanya data yang diberikan dan jangan membuat nilai yang tidak tersedia.
- Bedakan hasil pengolahan data dengan interpretasi geologi.
- Jangan langsung menetapkan jenis batuan hanya berdasarkan nilai kecepatan tanpa mempertimbangkan informasi pendukung.
- Jika terdapat lebih dari satu kemungkinan model, jelaskan alternatif tersebut dan tingkat ketidakpastiannya.
- Pertahankan satuan yang digunakan pada data atau jelaskan apabila dilakukan konversi satuan.

#### **O — Output (Keluaran)**

Hasil analisis harus mencakup:
1. Deskripsi data dan pola waktu tiba.
2. Analisis hubungan jarak dengan waktu tempuh.
3. Estimasi kecepatan gelombang pada setiap lapisan yang dapat diidentifikasi.
4. Perkiraan batas atau kedalaman lapisan jika data memungkinkan.
5. Interpretasi kondisi bawah permukaan.
6. Kesimpulan yang sesuai dengan data.

#### **Q — Quality Criteria (Kriteria Kualitas)**

1. Menggunakan prinsip metode seismik refraksi secara benar.
2. Perhitungan kecepatan dilakukan berdasarkan data yang tersedia.
3. Tidak mengarang nilai waktu tiba, kecepatan, kedalaman, atau informasi geologi.
4. Interpretasi dibedakan secara jelas dari hasil pengukuran.
5. Satuan dan posisi data konsisten.
6. Bahasa yang digunakan jelas, logis, dan sesuai tingkat mahasiswa S1 Geofisika.

#### **V — Verification (Verifikasi)**

Sebelum memberikan hasil akhir, lakukan pemeriksaan terhadap:
1. Kesesuaian posisi geofon dan jarak antar titik.
2. Konsistensi satuan jarak dan waktu.
3. Ketepatan perhitungan kecepatan.
4. Kesesuaian perubahan slope pada kurva travel time dengan interpretasi lapisan.
5. Kesesuaian interpretasi dengan informasi geologi yang tersedia.
6. Pastikan tidak ada nilai yang dibuat-buat.

### **PROMPT 2 : ANALISIS DATA GEOMAGNETIK**

#### **Struktur Prompt**

Judul: Analisis Data Geomagnetik untuk Mengidentifikasi Variasi Medan Magnet dan Struktur Bawah Permukaan

#### **R — Role (Peran)**

Bertindak sebagai ahli Geofisika terapan dan dosen pembimbing praktikum yang memiliki keahlian dalam metode geomagnetik. Lakukan analisis secara ilmiah dan sesuai tingkat pemahaman mahasiswa S1 Geofisika. Memiliki kemampuan dalam pengolahan data medan magnet, koreksi data, analisis anomali magnetik, pemetaan pola anomali, dan interpretasi struktur bawah permukaan.

#### **T — Task (Tugas)**

1. Analisis data geomagnetik untuk mengidentifikasi variasi nilai medan magnet pada daerah penelitian.
2. Jelaskan pola anomali magnetik yang terlihat pada data atau peta.
3. Identifikasi zona anomali relatif tinggi dan rendah.
4. Interpretasikan kemungkinan hubungan pola anomali dengan variasi litologi atau struktur geologi.
5. Susun kesimpulan berdasarkan data dan hasil interpretasi.

#### **C — Context (Konteks)**

Metode geomagnetik memanfaatkan variasi medan magnet bumi yang dipengaruhi oleh sifat kemagnetan material bawah permukaan. Perbedaan susceptibilitas magnetik dan keberadaan material bermagnet dapat menghasilkan anomali magnetik. Interpretasi anomali harus mempertimbangkan posisi, pola, amplitudo, arah perubahan anomali, kondisi geologi, serta proses pengolahan data yang telah dilakukan.

#### **I — Input (Masukan)**

1. Data hasil pengukuran medan magnet.
2. Koordinat setiap titik pengukuran.
3. Nilai medan magnet atau nilai anomali magnetik.
4. Peta kontur atau peta anomali magnetik jika tersedia.
5. Informasi geologi dan topografi daerah penelitian jika tersedia.

#### **K — Constraints (Batasan)**

- Gunakan hanya data yang diberikan dan jangan mengarang nilai.
- Jelaskan jenis koreksi yang telah dilakukan apabila informasinya tersedia; jangan mengasumsikan koreksi tertentu telah dilakukan.
- Bedakan pola anomali yang merupakan hasil data dengan interpretasi geologi.
- Jangan menetapkan jenis batuan atau struktur tertentu hanya dari satu pola anomali tanpa data pendukung.
- Jika interpretasi memiliki beberapa kemungkinan, jelaskan sebagai kemungkinan dan sertakan keterbatasannya.

#### **O — Output (Keluaran)**

Hasil analisis harus mencakup:
1. Deskripsi karakteristik data geomagnetik.
2. Identifikasi zona anomali tinggi dan rendah.
3. Analisis pola spasial anomali.
4. Interpretasi kemungkinan variasi litologi atau struktur bawah permukaan.
5. Hubungan hasil geomagnetik dengan informasi geologi yang tersedia.
6. Kesimpulan berdasarkan hasil analisis.

#### **Q — Quality Criteria (Kriteria Kualitas)**

1. Menggunakan konsep dasar metode geomagnetik dengan benar.
2. Analisis hanya menggunakan data yang tersedia.
3. Tidak membuat nilai medan magnet, anomali, atau informasi geologi yang tidak diberikan.
4. Interpretasi anomali didukung oleh pola data dan informasi pendukung.
5. Ketidakpastian interpretasi dijelaskan dengan jelas.
6. Hasil ditulis dengan bahasa ilmiah, sistematis, dan sesuai tingkat mahasiswa S1 Geofisika.

#### **V — Verification (Verifikasi)**

1. Periksa kesesuaian koordinat dan posisi titik pengukuran.
2. Periksa konsistensi satuan medan magnet atau anomali yang digunakan.
3. Pastikan zona anomali yang disebutkan benar-benar terdapat pada data atau peta.
4. Pastikan interpretasi tidak melebihi informasi yang dapat didukung oleh data.
5. Bandingkan interpretasi dengan informasi geologi atau data pendukung jika tersedia.
6. Pastikan kesimpulan menjawab tujuan analisis geomagnetik.
